```python
import pandas as pd

data = pd.read_csv("DisasterDeclarationsSummaries.csv")

data.head()
```

```python
data.info()
data.describe()
```

```python
data.isnull().sum()
```

```python
data = data[['state', 'incidentType', 'declarationDate']]
data['declarationDate'] = pd.to_datetime(data['declarationDate'], errors='coerce')
data['year'] = data['declarationDate'].dt.year
data['month'] = data['declarationDate'].dt.month
```

```python
data = data.dropna()
```

```python
data = data.drop_duplicates()
```

```python
data.to_csv("usnd_cleaned.csv", index=False)
```
