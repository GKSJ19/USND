```python
import pandas as pd
df = pd.read_csv('/content/database.csv')
print("First 5 rows of the dataset:")
display(df.head())
print("\nDataFrame Information:")
df.info()
```

```python
import plotly.express as px
declarations_per_state = df['State'].value_counts().reset_index()
declarations_per_state.columns = ['State', 'Total Declarations']

# Display the aggregated data
print("Total Declarations per State:")
display(declarations_per_state.head())

# Create a choropleth map
fig = px.choropleth(
    declarations_per_state,
    locations='State',
    locationmode='USA-states',
    color='Total Declarations',
    scope='usa',
    color_continuous_scale="Viridis",
    title='Total Number of Disaster Declarations Per State'
)

fig.show()
```

```python
print("Unique Disaster Types:")
print(df['Disaster Type'].unique())
```

```python
import plotly.express as px

selected_disaster_type_hurricane = 'Hurricane'

df_selected_disaster_hurricane = df[df['Disaster Type'] == selected_disaster_type_hurricane]

# Aggregate data: total declarations per state for the selected disaster type
declarations_per_state_hurricane = df_selected_disaster_hurricane['State'].value_counts().reset_index()
declarations_per_state_hurricane.columns = ['State', 'Total Declarations']

print(f"Total {selected_disaster_type_hurricane} Declarations per State:")
display(declarations_per_state_hurricane.head())

# Create a choropleth map for the selected disaster type
fig_hurricane = px.choropleth(
    declarations_per_state_hurricane,
    locations='State',
    locationmode='USA-states',
    color='Total Declarations',
    scope='usa',
    color_continuous_scale="Viridis",
    title=f'Total Number of {selected_disaster_type_hurricane} Declarations Per State'
)

fig_hurricane.show()
```
