```python
# STEP 1: Import Required Libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set clean style
plt.style.use('default')
sns.set(font_scale=1.1)
```

```python
# STEP 2: Load Datasets
clean_df = pd.read_csv("../data/processed/usnd_cleaned.csv")
raw_df = pd.read_csv("../data/us_disaster_declarations.csv")

clean_df.head()
```

```python
# STEP 3: Prepare Raw Data
raw_df = raw_df[['state','incident_type','ih_program_declared','pa_program_declared']]

raw_df = raw_df.rename(columns={
    'incident_type': 'incidentType'
})

# Merge datasets
df = pd.merge(
    clean_df,
    raw_df,
    on=['state','incidentType'],
    how='left'
)
```

```python
# STEP 4: Top 10 Disaster Types

top_incidents = df['incidentType'].value_counts().head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=top_incidents.values, y=top_incidents.index)

plt.title("Top 10 Disaster Types", fontsize=14, weight='bold')
plt.xlabel("Number of Occurrences")
plt.ylabel("Disaster Type")

plt.tight_layout()
plt.show()
```

```python
# STEP 5: Aggregation

incident_state = df.groupby(['state','incidentType']).size().reset_index(name='count')

pivot_data = incident_state.pivot(
    index='state',
    columns='incidentType',
    values='count'
).fillna(0)
```

```python
# STEP 6: Heatmap

plt.figure(figsize=(14,8))
sns.heatmap(pivot_data, cmap="coolwarm")

plt.title("State vs Incident Type Heatmap", fontsize=14, weight='bold')
plt.xlabel("Disaster Type")
plt.ylabel("State")

plt.tight_layout()
plt.show()
```

```python
# STEP 7: Stacked Bar (Top States)

top_states = pivot_data.sum(axis=1).sort_values(ascending=False).head(10).index

pivot_data.loc[top_states].plot(
    kind='bar',
    stacked=True,
    figsize=(12,6)
)

plt.title("Disaster Type Distribution by State", fontsize=14, weight='bold')
plt.xlabel("State")
plt.ylabel("Disaster Count")

plt.legend(title="Disaster Type", bbox_to_anchor=(1.05,1))
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# STEP 8: Top 10 States

state_counts = df.groupby('state').size().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=state_counts.values, y=state_counts.index)

plt.title("Top 10 States with Highest Disaster Declarations", fontsize=14, weight='bold')
plt.xlabel("Number of Disasters")
plt.ylabel("State")

plt.tight_layout()
plt.show()
```

```python
# STEP 9: Monthly Distribution

monthly = df.groupby('month').size()

# Map month numbers to names
month_names = {
    1: "Jan", 2: "Feb", 3: "Mar", 4: "Apr",
    5: "May", 6: "Jun", 7: "Jul", 8: "Aug",
    9: "Sep", 10: "Oct", 11: "Nov", 12: "Dec"
}

monthly.index = monthly.index.map(month_names)

plt.figure(figsize=(10,5))
sns.barplot(x=monthly.index, y=monthly.values)

plt.title("Monthly Disaster Distribution", fontsize=14, weight='bold')
plt.xlabel("Month")
plt.ylabel("Number of Disasters")

plt.tight_layout()
plt.show()
```

```python
# STEP 10: Assistance Program Analysis

assistance = df.groupby('incidentType')[['ih_program_declared','pa_program_declared']].sum()

# Top 10 disaster types only (clean graph)
top_assistance = assistance.sort_values(by='pa_program_declared', ascending=False).head(10)

top_assistance.plot(kind='bar', figsize=(12,6))

plt.title("Top Disaster Types vs Assistance Programs", fontsize=14, weight='bold')
plt.xlabel("Disaster Type")
plt.ylabel("Assistance Count")

plt.xticks(rotation=45)
plt.legend(["Individual Housing", "Public Assistance"])

plt.tight_layout()
plt.show()
```
